```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# -----------------------------
# Sample Data (like a.csv)
# -----------------------------
data = {
    "age":    [25, 35, 45, 20, 50, 23, 42, 33],
    "income": [40, 60, 80, 30, 90, 35, 75, 55],
    "buys":   ["No", "Yes", "Yes", "No", "Yes", "No", "Yes", "Yes"]
}

df = pd.DataFrame(data)
print("Dataset:\n", df)

# Features and label
X = df[["age", "income"]]
y = df["buys"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1
)

# Model
nb = GaussianNB()
nb.fit(X_train, y_train)

# Predictions
y_pred = nb.predict(X_test)

# Accuracy
print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# -----------------------------
# Test with custom samples
# -----------------------------
test_sample = pd.DataFrame([[30, 50]], columns=["age", "income"])
print("Prediction for new sample:", nb.predict(test_sample))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# Load CSV
df = pd.read_csv("a.csv")

print(df.head())   # check the data

# Change 'label' to the actual name of your class column
label_col = "label"

# Split into features and target
X = df.drop(columns=[label_col])
y = df[label_col]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Model
nb = GaussianNB()
nb.fit(X_train, y_train)

# Predictions
y_pred = nb.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Custom test samples (optional)
# new = pd.DataFrame([[value1, value2, ...]], columns=X.columns)
# print("Prediction for custom test:", nb.predict(new))
```
